#exercicio letra B - Receber uma variavel, guarda-la e exibi-la

```python
nome = input("Qual seu nome?") #criação de varial "nome", "input" pra receber a variavel, "Qual seu nome", diz a caracteristica da variavel
idade = input("Quantos anos você tem?") #criação de varial "idade", "input" pra receber a variavel, "Qual sua idade", diz a caracteristica da variavel

print(f"{nome} Seu nome é muito bonito!") #Exibe os dados dentro das variaveis(nome), exibe essa variavel que teve a entrada recebida com uma pequena frase
print(f"Você tem {idade} anos!" ) #Exibe os dados dentro das variaveis(idade), exibe essa variavel que teve a entrada recebida com uma pequena frase
```

#exercicio letra C - Criação de duas variaveis Númericas, calculando soma e subtração das mesmas recebidas

```python
num1 = int(input("Escolha um número")) #criação variavel de um número, declaração dessa variavel como "int" pra declarar como um número na variavel, solicitando o número de entrada
num2 = int(input("Escolha outro número")) #criação variavel de outro número, declaração dessa variavel como "int" pra declarar como um número na variavel, solicitando o número de entrada
resultadosoma = num1 + num2 #criação variavel de somatoria das duas variaveis anteriores
resultadosubtração = num1 - num2 #criação variavel da subtração das duas variaveis anteriores

print(f"A soma dos Números escolhidos é {resultadosoma}") #Exibição da variável de soma com micro frase de descrição
print(f"A subtração dos números escolhidos é {resultadosubtração}") #Exibição da variável de subtração com micro frase de descrição



```
