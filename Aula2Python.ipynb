Exercicio 2 - Termostato com alerta de procedimentos com base em intervalos de temperatura

```python
#caso queira testar apenas o conhecimento
temperatura = 17 #variavel que afirma que a temperatura é 17

if temperatura <= 15: #caso a temperatura seja 0 a/ou 15 está frio
  print("A temperatura está muito baixa, use casacos e tome bastante aguá, o tempo seco pode causar irritação. cuidado!")
elif temperatura <= 25: #se o caso anterior não corespondeu e a temperatura é 16 a/ou 25, então está normal
  print("A temperatura está relativamente normal, tome cuidado e aproveite para se exercitar!")
elif temperatura <= 30: #se o anterior de 25 não se encaixa, os valores de 26 a/ou 30 serão altos
  print("A temperatura está muito alta, você precisa tomar aguá regularmente!")
else: #caso não encaixe em nada, só pode ser maior de 31, o que significa preocupação térmica
  print("Você corre risco de vida real, por favor tome cuidado!")
```

Exercicio 3 - Determinação de abordagem ao usuario em virtude do horario

```python
#Caso quira perguntar o horário
horario = float(input("Que Horas são?")) #varial de relogio pra determinar o valor de entrada do dispositivo, para saber a abordagem

if horario <= 4.00: #de 00.00 a/ou 04.00 está de madrugada
  print("Ta tirando no blhg lek, vai dormir")
elif horario <= 12.00: #de 4.01 a/ou 12.00 está de manhã
  print("Bom dia!")
elif horario <= 18.00: #de 12.01 a/ou 18.00 está de tarde
  print("Boa tarde!")
else: #em qualquer outro valor diferente aos de cima está de noite(18.01 a/ou 24.00)
  print("Boa noite")
```
